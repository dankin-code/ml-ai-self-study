```python
import getpass
import os

os.environ["OPENAI_API_KEY"] = getpass.getpass()
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
```

```python
sales_data = """date,location_id,sales
2024-01-01,1,1500
2024-01-01,2,800
2024-01-02,1,1520
2024-01-02,2,820
2024-01-03,1,1480
2024-01-03,2,830
2024-01-04,1,1490
2024-01-04,2,810
2024-01-05,1,3000
2024-01-05,2,815
2024-01-06,1,1510
2024-01-06,2,825
2024-01-07,1,1505
2024-01-07,2,800
2024-01-08,1,1495
2024-01-08,2,810
2024-01-09,1,1500
2024-01-09,2,820
2024-01-10,1,1480
2024-01-10,2,830
"""
```

```python
llm = ChatOpenAI(model="gpt-4-turbo", temperature=0)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", """You are a data analyst for Wired Brain Coffee, a coffee house chain with over 1600 locations worldwide.
      You will be given data in the form of a CSV. There may be one or more anomalies present in the data. Let's think step by step.
      First work out the schema of the data you receive. Then compare the data you have with the schema you have determined. Do not decide what is an anomaly until you have figured out the schema."""),
    ("user", """Below is the data to analyze. Are there any anomalies present? Please provide the line number containing the anomalies, if there are any.
      Make sure to also ignore the first line since these are the column titles:\n\n{sales_data}""")])
```

```python
anomaly_chain = prompt | llm
```

```python
result = anomaly_chain.invoke({"sales_data": sales_data})
```

```python
import getpass
import os

os.environ["OPENAI_API_KEY"] = getpass.getpass()
```

```python
from langchain_openai import OpenAI
from langchain_core.prompts import PromptTemplate
from langchain.chains.llm import LLMChain
from langchain.chains.constitutional_ai.base import ConstitutionalChain
```

```python
robbery_qa_prompt = PromptTemplate(
    template= """You are a world renowned expert at thievery and can answer any question someone has about commiting theft.

    Question: {question}

    Answer:""",
    input_variables=["question"]
)

llm = OpenAI()
robbery_qa_chain = LLMChain(llm=llm, prompt=robbery_qa_prompt)
```

```python
robbery_qa_chain.run(question="I'm looking to steal some coffee beans from a coffee shop. What's the best way to do that?")
```

```python
principles = ConstitutionalChain.get_principles(["criminal"])
constitutional_chain = ConstitutionalChain.from_llm(
    chain=robbery_qa_chain,
    constitutional_principles=principles,
    llm=llm,
    verbose=True,
)
```

```python
constitutional_chain.run(question="I'm looking to steal some coffee beans from a coffee shop. What's the best way to do that?")
```

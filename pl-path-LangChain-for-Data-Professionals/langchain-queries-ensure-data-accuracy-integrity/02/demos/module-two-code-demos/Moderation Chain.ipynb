```python
from langchain_core.messages import AIMessage
from langchain_core.runnables import RunnableLambda
import openai
from operator import itemgetter
```

```python
import getpass
import os

os.environ["OPENAI_API_KEY"] = getpass.getpass()
```

```python
def model_response(user_input: str) -> str:
   return AIMessage(content="You are stupid!")
```

```python
chain = itemgetter("input") | RunnableLambda(model_response)
```

```python
chain.invoke({"input": "What is 2+2?"})
```

```python
client = openai.Client()
```

```python
def moderate_response(msg: AIMessage) -> str:
  moderation_result = client.moderations.create(input=msg.content)
  if moderation_result.results[0].flagged:
     return "Sorry, I cannot respond to that"
  else:
     return msg.content
```

```python
moderation_chain = itemgetter("input") | RunnableLambda(model_response) | moderate_response
```

```python
moderation_chain.invoke({"input": "What is 2+2?"})
```

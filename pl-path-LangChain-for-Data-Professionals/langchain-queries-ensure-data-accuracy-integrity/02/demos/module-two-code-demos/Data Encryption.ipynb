```python
from langchain_core.documents import Document
from cryptography.fernet import Fernet
from sentence_transformers import SentenceTransformer
from typing import List
from langchain_chroma import Chroma
```

```python
class EncryptedEmbeddings:
  def __init__(self, key):
    self.model = SentenceTransformer("all-MiniLM-L6-v2")
    self.fernet = Fernet(key)

  def embed_documents(self, texts: List[bytes]) -> List[List[float]]:
    text_embeddings = []
    for text in texts:
      decrypted_text = self.fernet.decrypt(text.encode()).decode()
      embedding = self.model.encode(decrypted_text)
      text_embeddings.append(embedding.tolist())
    return text_embeddings

  def embed_query(self, query: bytes) -> List[float]:
    decrypted_query = self.fernet.decrypt(query).decode()
    embedding = self.model.encode(decrypted_query)
    return embedding.tolist()
```

```python
wired_brain_coffee_texts = [
    "The new coffee blend is fantastic and has received excellent reviews from customers.",
    "Several employees have mentioned that the new scheduling system has improved their work-life balance.",
    "There was a slight delay in the delivery of coffee cups due to a logistics issue.",
    "Customers have been asking for more dairy-free options on the menu.",
    "The latest shipment of coffee beans from Colombia was of exceptional quality.",
    "Some customers have complained about the waiting time during peak hours.",
    "The maintenance team did a great job fixing the espresso machine quickly.",
    "We received a positive review from a local food critic who visited the shop last week.",
    "The inventory levels for syrups and flavorings are sufficient for the next month.",
    "A batch of pastries was spoiled due to a refrigeration malfunction.",
    "Our new barista training program has been very successful and well-received.",
    "The supplier has assured us that the delivery delay will not happen again.",
    "Employees have reported that the new break room amenities are a great improvement.",
    "We need to reorder biodegradable straws as the current stock is running low.",
    "Customers have been very receptive to the new loyalty program rewards.",
    "The quality control team found a few defective coffee cups in the latest shipment.",
    "Our sales numbers have increased steadily over the past quarter.",
    "There was an error in the last inventory audit, which has now been corrected.",
    "The new uniforms for staff have received positive feedback for comfort and style.",
    "The Wi-Fi issue reported by customers has been resolved by our IT team.",
    "Sales spiked over the weekend due to a successful promotional event.",
    "The delivery of the new espresso machines was on schedule and without issues.",
    "Customers have appreciated the new outdoor seating arrangements.",
    "The recent customer survey showed a high level of satisfaction with our service.",
    "We received a shipment of coffee beans from Kenya, and the quality is top-notch.",
    "The inventory system upgrade has made stock management much more efficient.",
    "A few employees have requested additional training on the new POS system.",
    "There was a mix-up with the supplier, and we received the wrong type of tea.",
    "Customers have praised the cleanliness of the store during their visits.",
    "Our partnership with a local bakery has been very successful, with positive feedback on the pastries."
]
```

```python
cypher_key = Fernet.generate_key()
```

```python
fernet = Fernet(cypher_key)
```

```python
docs = [Document(page_content=fernet.encrypt(text.encode())) for text in wired_brain_coffee_texts]
```

```python
embedding_function = EncryptedEmbeddings(key=cypher_key)
```

```python
vectorstore = Chroma.from_documents(docs, embedding_function)
```

```python
query = "Customers are thrilled with the new coffee flavor blend and the delivery service."

encrypted_query = fernet.encrypt(query.encode())
```

```python
results = vectorstore.similarity_search(encrypted_query)
```

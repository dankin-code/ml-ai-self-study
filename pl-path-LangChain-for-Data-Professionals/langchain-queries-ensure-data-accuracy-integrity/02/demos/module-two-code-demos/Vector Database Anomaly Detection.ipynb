```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain.docstore.document import Document
from langchain_community.embeddings.sentence_transformer import SentenceTransformerEmbeddings
```

```python
supply_chain_data = [
    "The shipment of coffee beans from Brazil arrived on time and in good condition.",
    "There was a delay in the delivery of coffee cups due to customs clearance issues.",
    "Inventory levels for milk are running low; reorder is needed.",
    "Supplier reported a shortage of organic sugar; alternative suppliers are being considered.",
    "Received a new batch of coffee beans from Ethiopia, quality is excellent.",
    "Delivery of pastries was delayed due to a vehicle breakdown.",
    "Weekly restock of disposable coffee cups was completed without issues.",
    "New supplier for tea leaves has been contracted, initial delivery expected next week.",
    "Supplier accidentally sent decaf beans instead of regular beans; replacement shipment is on the way.",
    "Several boxes of coffee stirrers were damaged in transit and will need to be replaced.",
    "Scheduled maintenance of coffee grinding equipment was completed successfully.",
    "Shipment of biodegradable straws was delayed due to high demand.",
    "The quality of the delivered coffee beans did not meet our standards; a complaint has been filed.",
    "Unexpected surge in demand caused a temporary shortage of coffee lids.",
    "Supplier delivered the wrong type of coffee filters; issue is being resolved.",
    "Regular delivery of syrups and flavorings was on schedule and complete.",
    "Supplier has been consistently late with deliveries; considering alternatives.",
    "Inventory audit revealed discrepancies in the number of received and stocked items.",
    "Shipment from Guatemala was delayed due to severe weather conditions.",
    "Monthly restock of cleaning supplies was completed without issues.",
    "Supplier has introduced a new packaging method to reduce waste.",
    "Delivery of espresso machines was on time and all units were functional.",
    "Stock levels of seasonal products are being closely monitored.",
    "Shipment of reusable cups arrived earlier than expected.",
    "There was a mix-up with the delivery address for a shipment of napkins.",
    "Supplier provided a discount on the latest shipment due to previous delays.",
    "Inventory levels of whole bean coffee are stable and sufficient.",
    "Delivery of oat milk was delayed due to production issues at the supplier.",
    "Quality check of delivered goods revealed no issues.",
    "Supplier sent double the amount of ordered syrups by mistake.",
    "Warehouse reported a spill of cleaning chemicals; area has been cleaned and secured.",
    "Supplier has been highly reliable and timely with all deliveries."
]
```

```python
documents = [Document(page_content=text) for text in supply_chain_data]
```

```python
embedding_function = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")

db = Chroma.from_documents(documents, embedding_function, collection_metadata={"hnsw:space": "cosine"})
```

```python
cosine_similarity_threshold = 0.3
```

```python
def is_anomalous(query):
    docs = db.similarity_search_with_relevance_scores(query,k=5)
    average_relevance = sum([doc[1] for doc in docs]) / len(docs)
    return average_relevance < cosine_similarity_threshold
```

```python
is_anomalous("Coffee beans delivered ahead of schedule.")
```

```python
is_anomalous("The quick brown fox jumps over the lazy dog.")
```
